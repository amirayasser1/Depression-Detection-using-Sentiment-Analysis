```python
# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Set your folder path
base_path = "/content/drive/MyDrive/feature_extraction_outputs"

# Load data
import joblib

X_train = joblib.load(f"{base_path}/X_train_resampled.pkl")
y_train = joblib.load(f"{base_path}/y_train_resampled.pkl")

X_test  = joblib.load(f"{base_path}/X_test_combined.pkl")
y_test  = joblib.load(f"{base_path}/y_test.pkl")

print("Data loaded successfully!")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.kernel_approximation import RBFSampler
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, classification_report

# Model
kernel_lr = Pipeline([
    ("rbf", RBFSampler(gamma=0.1, random_state=42)),
    ("lr", LogisticRegression(max_iter=1000))
])

# Train
kernel_lr.fit(X_train, y_train)

# Predict
y_pred = kernel_lr.predict(X_test)

# Accuracy
acc = accuracy_score(y_test, y_pred)

print("=== Kernel Logistic Regression ===")
print("Accuracy:", acc)
print(classification_report(y_test, y_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report

# Model
rf = RandomForestClassifier(
    n_estimators=200,
    class_weight="balanced",
    random_state=42,
    n_jobs=-1
)

# Train
rf.fit(X_train, y_train)

# Predict
y_pred = rf.predict(X_test)

# Accuracy
acc = accuracy_score(y_test, y_pred)

print("=== Random Forest ===")
print("Accuracy:", acc)
print(classification_report(y_test, y_pred))
```
